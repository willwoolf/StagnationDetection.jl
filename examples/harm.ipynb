# Stagnation detection in practice

## Recursive summation

### Harmonic series

We provide a simple example for performing stagnation detection within a recursive sum.

The harmonic series is the sum that diverges in real arithmetic

$$ \sum_{k = 1}^{\infty} \frac{1}{k} $$

We can evaluate the harmonic series up to a fixed number of terms using the following Julia code

```julia
n = 4_000_000
acc = 0.0

for k in 1:n
    acc += 1/k
end

acc
```

First, we will express this as a function, allowing us to specify types

```julia
function harm(n::Int, T::Type = Float64)
    acc::T = 0.0
    for k in 1:n
        acc += one(T)/k
    end
    return acc
end
```

Then we can customise the format and the size of the summation in the function call

```julia
println(harm(2_000_000))
println(harm(3_000_000))
println(harm(4_000_000))
println(harm(2_000_000, Float32))
println(harm(3_000_000, Float32))
println(harm(4_000_000, Float32))
```

To implement stagnation detection, we can simply call `harm` with one of our own types defined for stagnation detection, and it will automatically record what we need

```julia
using StagnationDetection

h = harm(4_000_000, Float32SD)
```

And we can use `report` to identify where stagnation is occurring.

```julia
report(h)
```
